```python
# import các thư viện cần thiết
import numpy as np
import matplotlib.pyplot as plt

# Tên 5 lần huấn luyện và cấu hình tham số
model_names = [
    'Default\nRaw Data',
    'Default\nPreprocessed Data',
    'max_depth: None\nmin_samples_leaf: 2\nn_estimators: 150',
    'max_depth: 30\nmax_features: 0.8\nmin_samples_leaf: 2\nn_estimators: 200',
    'max_depth: 30\nmax_features: 0.6\nmin_samples_leaf: 1\nn_estimators: 300\nmin_samples_split: 2'
]

# Dữ liệu RMSE - Train
rmse_train_values = [
    203.3295,
    205.0551,
    271.1829,
    275.4599,
    199.1147
]

# Dữ liệu RMSE - Test
rmse_test_values = [
    533.0261,
    522.1064,
    520.3564,
    517.5292,
    521.0195
]

# Dữ liệu R² - Train
r2_train_values = [
    0.9974,
    0.9974,
    0.9954,
    0.9952,
    0.9975
]

# Dữ liệu R² - Test
r2_test_values = [
    0.9818,
    0.9828,
    0.9829,
    0.9831,
    0.9828
]

# Khởi tạo vị trí các mô hình trên trục hoành
x = np.arange(len(model_names))

# Khai báo độ rộng của mỗi cột
width = 0.2

# Khởi tạo biểu đồ với trục tung thứ nhất
fig, ax1 = plt.subplots(figsize=(14, 7))

# Vẽ cột RMSE trên tập Train - Xanh dương nhạt
bar1 = ax1.bar(
    x - 1.5 * width,
    rmse_train_values,
    width,
    label='RMSE Train',
    color='#89C2F5'
)

# Vẽ cột RMSE trên tập Test - Xanh dương đậm
bar2 = ax1.bar(
    x - 0.5 * width,
    rmse_test_values,
    width,
    label='RMSE Test',
    color='#2166AC'
)

# Đặt tên trục tung bên trái
ax1.set_ylabel('RMSE')

# Đặt tên trục hoành
ax1.set_xlabel('Model')

# Hiển thị nhãn các mô hình
ax1.set_xticks(x)
# Hiển thị nhãn các mô hình với font chữ nhỏ hơn
ax1.set_xticklabels(model_names, fontsize=9)

# Đặt giới hạn trục RMSE
ax1.set_ylim(0, 600)

# Tạo trục tung thứ hai
ax2 = ax1.twinx()

# Vẽ cột R² trên tập Train - Cam nhạt
bar3 = ax2.bar(
    x + 0.5 * width,
    r2_train_values,
    width,
    label='R² Train',
    color='#FFBE7A'
)

# Vẽ cột R² trên tập Test - Cam đậm
bar4 = ax2.bar(
    x + 1.5 * width,
    r2_test_values,
    width,
    label='R² Test',
    color='#E57222'
)

# Đặt tên trục tung bên phải
ax2.set_ylabel('R² Score')

# Đặt giới hạn trục R²
ax2.set_ylim(0.8, 1.01)

# Đặt tiêu đề biểu đồ
plt.title(
    'Comparison of RMSE and R² - Random Forest Regressor (Hold-out)',
    pad=55
)

# Kết hợp chú thích của hai trục
bars = [bar1, bar2, bar3, bar4]

# Đặt nhãn cho các cột
labels = [
    'RMSE Train',
    'RMSE Test',
    'R² Train',
    'R² Test'
]

# Đặt chú thích phía trên biểu đồ
ax1.legend(
    bars,
    labels,
    loc='upper center',
    bbox_to_anchor=(0.5, 1.17),
    ncol=4,
    frameon=False
)

# Hiển thị lưới ngang
ax1.grid(
    axis='y',
    linestyle='--',
    alpha=0.5
)

# Đặt lưới phía sau các cột
ax1.set_axisbelow(True)

# Điều chỉnh khoảng trống phía trên biểu đồ
plt.subplots_adjust(top=0.82)

# Điều chỉnh bố cục biểu đồ
plt.tight_layout()

# Hiển thị biểu đồ
plt.show()
```
