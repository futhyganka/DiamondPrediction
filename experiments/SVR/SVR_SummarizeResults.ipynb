```python

import numpy as np
import matplotlib.pyplot as plt

# Tên 5 lần huấn luyện và cấu hình tham số
model_names = [
    'Default\nRaw Data',
    'Default\nPreprocessed Data',
    'kernel: rbf\nC: 500\nepsilon: 1\ngamma: scale',
    'kernel: rbf\nC: 500\nepsilon: 0.5\ngamma: auto',
    'kernel: rbf\nC: 1000\nepsilon: 0.05\ngamma: 0.1'
]

# Dữ liệu RMSE - Train
rmse_train_values = [
    4241.0730,
    727.3570,
    625.7064,
    625.7145,
    599.9616
]

# Dữ liệu RMSE - Test
rmse_test_values = [
    4172.1029,
    748.1001,
    638.3387,
    638.3387,
    611.4455
]

# Dữ liệu R² - Train
r2_train_values = [
    -0.1249,
    0.9667,
    0.9754,
    0.9754,
    0.9774
]

# Dữ liệu R² - Test
r2_test_values = [
    -0.1149,
    0.9646,
    0.9742,
    0.9742,
    0.9763
]

# Khởi tạo vị trí các mô hình trên trục hoành
x = np.arange(len(model_names))

# Khai báo độ rộng của mỗi cột
width = 0.2

# Khởi tạo biểu đồ với trục tung thứ nhất
fig, ax1 = plt.subplots(figsize=(14, 7))

# Vẽ cột RMSE trên tập Train - Xanh dương nhạt
bar1 = ax1.bar(
    x - 1.5 * width,
    rmse_train_values,
    width,
    label='RMSE Train',
    color='#89C2F5'
)

# Vẽ cột RMSE trên tập Test - Xanh dương đậm
bar2 = ax1.bar(
    x - 0.5 * width,
    rmse_test_values,
    width,
    label='RMSE Test',
    color='#2166AC'
)

# Đặt tên trục tung bên trái
ax1.set_ylabel('RMSE')

# Đặt tên trục hoành
ax1.set_xlabel('Model')

# Hiển thị nhãn các mô hình
ax1.set_xticks(x)
ax1.set_xticklabels(model_names, fontsize=9)

# Đặt giới hạn trục RMSE
ax1.set_ylim(0, 4500)

# Tạo trục tung thứ hai
ax2 = ax1.twinx()

# Vẽ cột R² trên tập Train - Cam nhạt
bar3 = ax2.bar(
    x + 0.5 * width,
    r2_train_values,
    width,
    label='R² Train',
    color='#FFBE7A'
)

# Vẽ cột R² trên tập Test - Cam đậm
bar4 = ax2.bar(
    x + 1.5 * width,
    r2_test_values,
    width,
    label='R² Test',
    color='#E57222'
)

# Đặt giới hạn trục RMSE
ax1.set_ylim(-900, 4500)

# Đặt giới hạn trục R²
ax2.set_ylim(-0.2, 1.0)

# Đường tham chiếu tại giá trị 0
ax1.axhline(
    y=0,
    color='gray',
    linestyle='--',
    linewidth=1,
    alpha=0.7
)

# Đặt tiêu đề biểu đồ
plt.title(
    'Comparison of RMSE and R² - SVR (Hold-out)',
    pad=55
)

# Kết hợp chú thích của hai trục
bars = [bar1, bar2, bar3, bar4]

labels = [
    'RMSE Train',
    'RMSE Test',
    'R² Train',
    'R² Test'
]

# Đặt chú thích phía trên biểu đồ
ax1.legend(
    bars,
    labels,
    loc='upper center',
    bbox_to_anchor=(0.5, 1.17),
    ncol=4,
    frameon=False
)

# Hiển thị lưới ngang
ax1.grid(
    axis='y',
    linestyle='--',
    alpha=0.5
)

ax1.set_axisbelow(True)

# Hiển thị đường tham chiếu R² = 0
ax2.axhline(
    y=0,
    color='gray',
    linestyle=':',
    linewidth=1,
    alpha=0.7
)

# Điều chỉnh khoảng trống phía trên biểu đồ
plt.subplots_adjust(top=0.82)

# Điều chỉnh bố cục biểu đồ
plt.tight_layout()

# Hiển thị biểu đồ
plt.show()
```
