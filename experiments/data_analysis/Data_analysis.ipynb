```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download(
    "nancyalaswad90/diamonds-prices",
    output_dir="/kaggle/input/"
    )

print("Path to dataset files:", path)
```

```python
import pandas as pd

df = pd.read_csv("/kaggle/input/diamonds-prices/Diamonds Prices2022.csv", sep = ",", index_col = 0)

print("5 Dữ liệu đầu:\n", df.head())
print("(Số thuộc tính, số dữ liệu):", df.shape)
print("Các thuộc tính:", df.info())
print("Cụ thể từng thuộc tính:\n", df.describe())
print("\ncut:", df['cut'].value_counts())
print("\ncolor", df['color'].value_counts())
print("", df['clarity'].value_counts())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='price', bins=50)
plt.title('Biểu đồ phân phối giá kim cương')
plt.xlabel('Giá (USD)')
plt.ylabel('Số lượng')
plt.show()
```
